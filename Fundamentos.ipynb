## Metodo print

### Exemplos 1
##### Parametro %s é utiizado para receber informações de fora da celula


```python
# Variáveis
nome = "joão"
idade = 25
end = " Rua 31 Norte lote 04"
cep = 71918360
tel = "(61) 984575053"

print ("nome: %s idade: %s endereço: %s CEP: %s tel: %s" % (nome, idade, end, cep, tel))
```

### Exemplos 2
###### Parametro {} + .format é utiizado  da mesma maneira do %s para receber informações de fora da celula


```python
# Variáveis
nome = "joão"
idade = 25
end = " Rua 31 Norte lote 04"
cep = 71918360
tel = "(61) 984575053"

print ("nome: {} idade: {} endereço: {} CEP: {} tel: {}" .format (nome, idade, end, cep, tel))
```

### Exemplo 3
###### Parametro sep="" serve para colocar qualquer tipo de separador entre os valores

```python
print ("Rua", 31, "Norte", "lote", 4.14,  sep=" - ")
print ("Rua", 31, "Norte", "lote", 4.14,  sep=" / ")
print ("Rua", 31, "Norte", "lote", 4.14,  sep=" > ")
print ("Rua", 31, "Norte", "lote", 4.14,  sep=" || ")
```

### Exemplo 4
#####Utilizando o Parametro [end]  para "fim" ao final da celula


```python
print ("minha", "Casa", "Amarela", sep=' - ', end=' Fim')
```

### Exemplo 5
Utilizando o 'f' string


```python
nome = "renan"
print (f"Nome:{nome}")
```
